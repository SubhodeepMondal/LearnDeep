```python
np.random.seed(63)
A_add_input_33_16 = np.random.rand(33, 16)
B_add_input_16 = np.random.rand(16)
E_add_input_33_16 = np.random.rand(33, 16)
print("Inputs:")

print(','.join(f'{x:.{8}f}' for x in A_add_input_33_16.flatten()))
print(','.join(f'{x:.{8}f}' for x in B_add_input_16.flatten()))
print(','.join(f'{x:.{8}f}' for x in E_add_input_33_16.flatten()))


C_33_16 = np.add(A_add_input_33_16, B_add_input_16)
D_33_16 = np.multiply(C_33_16, E_add_input_33_16)
F_33_16 = np.multiply(C_33_16, D_33_16)

print("Outputs")
print(','.join(f'{x:.{8}f}' for x in F_33_16.flatten()))

dF = np.ones_like(F_33_16)              # upstream gradient

# F = C * D
dC_from_F = dF * D_33_16
dD = dF * C_33_16

# D = C * E
dC_from_D = dD * E_add_input_33_16
dE = dD * C_33_16

# total gradient for C
dC = dC_from_F + dC_from_D

# C = A + B
dA = dC
dB = np.sum(dC, axis=0)

print("Gradients")
print(','.join(f'{x:.{8}f}' for x in dA.flatten()))
print(','.join(f'{x:.{8}f}' for x in dB.flatten()))
print(','.join(f'{x:.{8}f}' for x in dC.flatten()))
print(','.join(f'{x:.{8}f}' for x in dD.flatten()))
print(','.join(f'{x:.{8}f}' for x in dE.flatten()))
print(','.join(f'{x:.{8}f}' for x in dF.flatten()))
```

```python
np.random.seed(19)
A_input_32_42 = 3 * np.random.rand(32, 42)
B_input_42 = 4 * np.random.rand( 42)
E_input_42_72 = 2 * np.random.rand(42, 72)
F_input_32_42 = 5 * np.random.rand(32, 42)
print("Inputs")
print(','.join(f'{x:.{8}f}' for x in A_input_32_42.flatten()))
print(','.join(f'{x:.{8}f}' for x in B_input_42.flatten()))
print(','.join(f'{x:.{8}f}' for x in E_input_42_72.flatten()))
print(','.join(f'{x:.{8}f}' for x in F_input_32_42.flatten()))

C_32_42 = np.subtract(A_input_32_42, B_input_42)
D_32_42 = np.multiply(F_input_32_42, C_32_42)
G_32_72 = np.matmul(C_32_42, E_input_42_72)

print("Output")
print(','.join(f'{x:.{8}f}' for x in C_32_42.flatten()))
print(','.join(f'{x:.{8}f}' for x in D_32_42.flatten()))
print(','.join(f'{x:.{8}f}' for x in G_32_72.flatten()))

# gradient calculation
# ----- backward -----
dD = np.ones_like(D_32_42)          # ∂L/∂D
dG = np.ones_like(G_32_72)          # ∂L/∂G

# ---- D = F * C ----
dF = dD * C_32_42
dC_from_D = dD * F_input_32_42

# ---- G = C @ E ----
dC_from_G = dG @ E_input_42_72.T          # (32,72) @ (72,42)
dE = C_32_42.T @ dG                 # (42,32) @ (32,72)

# ---- accumulate gradients for C ----
dC = dC_from_D + dC_from_G

# ---- C = A - B ----
dA = dC
dB = -np.sum(dC, axis=0)      # reduce broadcast axis

print("Gradients")
print(','.join(f'{x:.{8}f}' for x in dA.flatten()))
print(','.join(f'{x:.{8}f}' for x in dB.flatten()))
print(','.join(f'{x:.{8}f}' for x in dC.flatten()))
print(','.join(f'{x:.{8}f}' for x in dE.flatten()))
print(','.join(f'{x:.{8}f}' for x in dF.flatten()))
```

```python
np.random.seed(32)


batch_size = 128

x_input = np.random.rand(batch_size, 43)
y_output = np.random.rand(batch_size, 32)
weights = np.random.rand(43,32)
bias = np.random.rand(32)

print('Inputs')
print(','.join(f'{x:.{8}f}' for x in x_input.flatten()))
print(','.join(f'{x:.{8}f}' for x in y_output.flatten()))
print(','.join(f'{x:.{8}f}' for x in weights.flatten()))
print(','.join(f'{x:.{8}f}' for x in bias.flatten()))

# --------------------
# forward
# --------------------
y_hat = np.matmul(x_input, weights) + bias

print('Output')
print(','.join(f'{x:.{8}f}' for x in y_hat.flatten()))

# --------------------
# (Loss)
# --------------------
error = y_hat - y_output

loss = loss = np.sum(error ** 2, axis=0) / batch_size   # scalar
print("Loss:")
print(','.join(f'{x:.8f}' for x in loss.flatten()))

# --------------------
# gradient
# --------------------
grad_loss = (2.0 / batch_size) * error      # dL/dY_hat
grad_weight = np.matmul(x_input.T, grad_loss)
grad_bias = np.sum(grad_loss, axis=0)


print("Gradients:")
print(','.join(f'{x:.8f}' for x in grad_loss.flatten()))
print(','.join(f'{x:.8f}' for x in grad_weight.flatten()))
print(','.join(f'{x:.8f}' for x in grad_bias.flatten()))

learning_rate = 0.01
updated_weight = weights - learning_rate * grad_weight
updated_bias = bias - learning_rate * grad_bias

print("Updated Weights bias")
print(','.join(f'{x:.8f}' for x in updated_weight.flatten()))
print(','.join(f'{x:.8f}' for x in updated_bias.flatten()))


```

## Test 2, DenseLayer_Loss_Integration_parameter_update_multiple_epochs

```python
from dataclasses import dataclass
from typing import Dict
import numpy as np


@dataclass
class DenseTest4SimulateEpoch:
    x: np.ndarray  # (sample_size, in_features)
    w: np.ndarray  # (in_features, dense_units)
    b: np.ndarray  # (dense_units,)
    y: np.ndarray  # (sample_size, dense_units)


def load_dense_test_4(
    sample_size: int,
    no_of_input_features: int,
    no_of_dense_units: int,
    seed: int = 77,
) -> DenseTest4SimulateEpoch:
    rng = np.random.default_rng(seed)
    x = rng.random((sample_size, no_of_input_features))
    w = rng.random((no_of_input_features, no_of_dense_units))
    b = rng.random((no_of_dense_units,))
    y = rng.random((sample_size, no_of_dense_units))
    return DenseTest4SimulateEpoch(x=x, w=w, b=b, y=y)


def simulate_training(
    data: DenseTest4SimulateEpoch,
    epochs: int,
    batch_size: int,
    learning_rate: float,
) -> Dict[str, np.ndarray]:
    x, w, b, y = data.x, data.w, data.b, data.y

    if epochs <= 0:
        raise ValueError("epochs must be > 0")
    if batch_size <= 0 or batch_size > x.shape[0]:
        raise ValueError(f"batch_size must be in [1, {x.shape[0]}]")
    if x.shape[0] != y.shape[0]:
        raise ValueError("x and y must have same sample size")

    sample_size = x.shape[0]
    num_batches = sample_size // batch_size  # drop remainder
    if num_batches == 0:
        raise ValueError("batch_size is too large; no full batch available")

    all_batch_index = []
    all_input_batch = []
    all_output_batch = []
    all_loss_batch = []
    all_grad_loss = []
    all_grad_weight = []
    all_grad_bias = []
    all_updated_weight = []
    all_updated_bias = []
    all_target_output = []

    curr_w = w.copy()
    curr_b = b.copy()

    # One batch per epoch-step, rolling over full batches in order.
    for ep in range(epochs):
        epoch_loss = 0.0
        for batch_idx in range(num_batches):
            start = batch_idx * batch_size
            end = start + batch_size

            x_batch = x[start:end, :]  # (batch_size, in_features)
            y_batch = y[start:end, :]  # (batch_size, dense_units)

            output = x_batch @ curr_w + curr_b
            diff = output - y_batch
            loss = np.mean(diff * diff, axis=0)
            epoch_loss += loss

            grad_loss = (2.0 / batch_size) * diff
            grad_weight = x_batch.T @ grad_loss
            grad_bias = np.sum(grad_loss, axis=0)

            updated_w = curr_w - learning_rate * grad_weight
            updated_b = curr_b - learning_rate * grad_bias
            print(loss.shape)

            all_batch_index.append(batch_idx)
            all_input_batch.append(x_batch.copy())
            all_output_batch.append(output.copy())
            all_loss_batch.append(loss.copy())
            all_grad_loss.append(grad_loss.copy())
            all_grad_weight.append(grad_weight.copy())
            all_grad_bias.append(grad_bias.copy())
            all_updated_weight.append(updated_w.copy())
            all_updated_bias.append(updated_b.copy())
            all_target_output.append(y_batch.copy())

            # step-wise SGD update continuity
            curr_w = updated_w
            curr_b = updated_b

        epoch_loss /= num_batches

    return {
        "batch_index": np.array(all_batch_index, dtype=np.int64),
        "input": np.stack(all_input_batch, axis=0),
        "output": np.stack(all_output_batch, axis=0),
        "target_output": np.stack(all_target_output, axis=0),
        "loss": np.stack(all_loss_batch, axis=0),
        "grad_loss": np.stack(all_grad_loss, axis=0),
        "grad_weight": np.stack(all_grad_weight, axis=0),
        "grad_bias": np.stack(all_grad_bias, axis=0),
        "updated_weight": np.stack(all_updated_weight, axis=0),
        "updated_bias": np.stack(all_updated_bias, axis=0),
    }

```

```python
data = load_dense_test_4(128, 17, 19)
all_data = simulate_training(data, epochs=10, batch_size=32, learning_rate=0.01)



data.x.astype(data.x.astype(np.float64).tofile("../data/DenseLayer_Test_2_input_data.bin"))
data.w.astype(data.w.astype(np.float64).tofile("../data/DenseLayer_Test_2_weights_data.bin"))
data.b.astype(data.b.astype(np.float64).tofile("../data/DenseLayer_Test_2_bias_data.bin"))
data.y.astype(data.y.astype(np.float64).tofile("../data/DenseLayer_Test_2_output_data.bin"))


all_data["input"].astype(all_data["input"].astype(np.float64).tofile("../data/DenseLayer_Test_2_input_batch_data.bin"))
all_data["output"].astype(all_data["output"].astype(np.float64).tofile("../data/DenseLayer_Test_2_output_batch_data.bin"))
all_data["target_output"].astype(all_data["target_output"].astype(np.float64).tofile("../data/DenseLayer_Test_2_target_output_batch_data.bin"))
all_data["loss"].astype(all_data["loss"].astype(np.float64).tofile("../data/DenseLayer_Test_2_loss_batch_data.bin"))
all_data["grad_loss"].astype(all_data["grad_loss"].astype(np.float64).tofile("../data/DenseLayer_Test_2_grad_loss_batch_data.bin"))
all_data["grad_weight"].astype(all_data["grad_weight"].astype(np.float64).tofile("../data/DenseLayer_Test_2_grad_weight_batch_data.bin"))
all_data["grad_bias"].astype(all_data["grad_bias"].astype(np.float64).tofile("../data/DenseLayer_Test_2_grad_bias_batch_data.bin"))
all_data["updated_weight"].astype(all_data["updated_weight"].astype(np.float64).tofile("../data/DenseLayer_Test_2_updated_weight_batch_data.bin"))
all_data["updated_bias"].astype(all_data["updated_bias"].astype(np.float64).tofile("../data/DenseLayer_Test_2_updated_bias_batch_data.bin"))
```

## Test 3, double dense layer convergance test

```python
np.random.seed(42)

# -----------------------
# Config
# -----------------------
N = 512
D_in, D_hidden, D_out = 27, 31, 7
batch_size = 256
epochs = 100
lr = 0.01
noise_level = 0.01

# -----------------------
# Data (2-layer ground truth)
# -----------------------
X = np.random.randn(N, D_in)

W1_true = np.random.randn(D_in, D_hidden)
b1_true = np.random.randn(D_hidden)

W2_true = np.random.randn(D_hidden, D_out)
b2_true = np.random.randn(D_out)

Z_true = X @ W1_true + b1_true
Y = Z_true @ W2_true + b2_true

# Add noise
Y += noise_level * np.random.randn(*Y.shape)

# -----------------------
# Model init
# -----------------------
W1 = 0.01 * np.random.randn(D_in, D_hidden)
b1 = np.zeros(D_hidden)


W2 = 0.01 * np.random.randn(D_hidden, D_out)
b2 = np.zeros(D_out)


W1_pretrainning = W1.copy()
b1_pretraining = b1.copy()
W2_pretrainning = W2.copy()
b2_pretraining = b2.copy()

all_batch_index = []
all_input = []
all_output = []
all_loss = []
all_grad_loss = []
all_grad_weight_1 = []
all_grad_bias_1 = []
all_updated_weights_1 = []
training_weights_1_batch = []
training_weights_2_batch = []
all_updated_bias_1 = []
all_grad_weights_2 = []
all_grad_bias_2 = []
all_updated_weights_2 = []
all_updated_bias_2 = []
all_target_output = []


# -----------------------
# Training
# -----------------------
num_batches = N // batch_size

for epoch in range(epochs):
    epoch_loss = 0.0

    for i in range(num_batches):
        training_weights_1_batch.append(W1.copy())
        training_weights_2_batch.append(W2.copy())
        
        s = i * batch_size
        e = s + batch_size

        Xb = X[s:e]
        Yb = Y[s:e]

        # Forward
        Z1 = Xb @ W1 + b1
        Yp = Z1 @ W2 + b2

        # Loss
        sample_loss = np.mean((Yp - Yb) ** 2, axis=1)
        loss = np.mean(sample_loss)
        epoch_loss += loss

        # Backward
        dY =  (2.0 / batch_size) * (Yp - Yb)

        dW2 = Z1.T @ dY
        db2 = np.sum(dY, axis=0)

        dZ1 = dY @ W2.T

        dW1 = Xb.T @ dZ1
        db1 = np.sum(dZ1, axis=0)

        # Update
        W1 -= lr * dW1
        b1 -= lr * db1
        W2 -= lr * dW2
        b2 -= lr * db2

        all_input.append(Xb.copy())
        all_updated_bias_1.append(b1.copy())
        all_updated_bias_2.append(b2.copy())
        all_grad_weights_2.append(dW2.copy())
        all_updated_weights_1.append(W1.copy())
        all_updated_weights_2.append(W2.copy())
        all_grad_loss.append(dY.copy())


    epoch_loss /= num_batches
    all_loss.append(epoch_loss.copy())

    if epoch % 5 == 0:
        print(f"Epoch {epoch:3d} | Loss: {epoch_loss:.6f}")

# -----------------------
# Verification
# -----------------------
Y_pred = (X @ W1 + b1) @ W2 + b2
print("\nFinal MSE:", np.mean((Y_pred - Y) ** 2))

# Effective weights check
W_eff = W1 @ W2
b_eff = b1 @ W2 + b2

W_true_eff = W1_true @ W2_true
b_true_eff = b1_true @ W2_true + b2_true

print("W error:", np.linalg.norm(W_eff - W_true_eff))
print("b error:", np.linalg.norm(b_eff - b_true_eff))
```

```python
X.astype(X.astype(np.float64).tofile("../data/DenseLayer_Test_3_input_sample.bin"))
W1_pretrainning.astype(W1_pretrainning.astype(np.float64).tofile("../data/DenseLayer_Test_3_weights_1.bin"))
b1_pretraining.astype(b1_pretraining.astype(np.float64).tofile("../data/DenseLayer_Test_3_bias_1.bin"))
W2_pretrainning.astype(W2_pretrainning.astype(np.float64).tofile("../data/DenseLayer_Test_3_weights_2.bin"))
b2_pretraining.astype(b2_pretraining.astype(np.float64).tofile("../data/DenseLayer_Test_3_bias_2.bin"))
Y.astype(Y.astype(np.float64).tofile("../data/DenseLayer_Test_3_target_output.bin"))

all_batch_input_nparray = np.array(all_input, dtype=np.float64)
all_updated_bias_1_nparray = np.array(all_updated_bias_1, dtype=np.float64)
all_updated_bias_2_nparray = np.array(all_updated_bias_2, dtype=np.float64)
all_updated_weights_1_nparray = np.array(all_updated_weights_1, dtype=np.float64)
all_updated_weights_2_nparray = np.array(all_updated_weights_2, dtype=np.float64)
all_grad_weights_2_nparray = np.array(all_grad_weights_2, dtype=np.float64)
training_weights_1_batch_nparray = np.array(training_weights_1_batch, dtype=np.float64)
training_weights_2_batch_nparray = np.array(training_weights_2_batch, dtype=np.float64)
all_loss_grad_nparray = np.array(all_grad_loss, dtype=np.float64)
all_loss_nparray = np.array(all_loss, dtype=np.float64)


all_batch_input_nparray.tofile("../data/DenseLayer_Test_3_batch_input_history.bin")
all_updated_bias_1_nparray.tofile("../data/DenseLayer_Test_3_b1_history.bin")
all_updated_bias_2_nparray.tofile("../data/DenseLayer_Test_3_b2_history.bin")
all_updated_weights_1_nparray.tofile("../data/DenseLayer_Test_3_w1_history.bin")
all_updated_weights_2_nparray.tofile("../data/DenseLayer_Test_3_w2_history.bin")
all_loss_grad_nparray.tofile("../data/DenseLayer_Test_3_loss_grad_batch_end_history.bin")
training_weights_1_batch_nparray.tofile("../data/DenseLayer_Test_3_w1_batch_begin_history.bin")
all_grad_weights_2_nparray.tofile("../data/DenseLayer_Test_3_w2_grad_batch_end_history.bin")
training_weights_2_batch_nparray.tofile("../data/DenseLayer_Test_3_w2_batch_begin_history.bin")
all_loss_nparray.tofile("../data/DenseLayer_Test_3_epoch_loss.bin")
```
