```python
np.random.seed(19)
A_input_32_42 = 3 * np.random.rand(32, 42)
B_input_42 = 4 * np.random.rand( 42)
E_input_42_72 = 2 * np.random.rand(42, 72)
F_input_32_42 = 5 * np.random.rand(32, 42)
print("Inputs")
print(','.join(f'{x:.{8}f}' for x in A_input_32_42.flatten()))
print(','.join(f'{x:.{8}f}' for x in B_input_42.flatten()))
print(','.join(f'{x:.{8}f}' for x in E_input_42_72.flatten()))
print(','.join(f'{x:.{8}f}' for x in F_input_32_42.flatten()))

C_32_42 = np.subtract(A_input_32_42, B_input_42)
D_32_42 = np.multiply(F_input_32_42, C_32_42)
G_32_72 = np.matmul(C_32_42, E_input_42_72)

print("Output")
print(','.join(f'{x:.{8}f}' for x in C_32_42.flatten()))
print(','.join(f'{x:.{8}f}' for x in D_32_42.flatten()))
print(','.join(f'{x:.{8}f}' for x in G_32_72.flatten()))

# gradient calculation
# ----- backward -----
dD = np.ones_like(D_32_42)          # ∂L/∂D
dG = np.ones_like(G_32_72)          # ∂L/∂G

# ---- D = F * C ----
dF = dD * C_32_42
dC_from_D = dD * F_input_32_42

# ---- G = C @ E ----
dC_from_G = dG @ E_input_42_72.T          # (32,72) @ (72,42)
dE = C_32_42.T @ dG                 # (42,32) @ (32,72)

# ---- accumulate gradients for C ----
dC = dC_from_D + dC_from_G

# ---- C = A - B ----
dA = dC
dB = -np.sum(dC, axis=0)      # reduce broadcast axis

print("Gradients")
print(','.join(f'{x:.{8}f}' for x in dA.flatten()))
print(','.join(f'{x:.{8}f}' for x in dB.flatten()))
print(','.join(f'{x:.{8}f}' for x in dC.flatten()))
print(','.join(f'{x:.{8}f}' for x in dE.flatten()))
print(','.join(f'{x:.{8}f}' for x in dF.flatten()))
```

### Softmax gradient Test 1

```python
import numpy as np

np.random.seed(13)


def softmax(x, axis):
    x_max = np.max(x, axis=axis, keepdims=True)
    exp_x = np.exp(x - x_max)
    return exp_x / np.sum(exp_x, axis=axis, keepdims=True)


# ============================================================
# Test 1: Standalone Softmax autograd
#
# Framework:
#
# A [52,162]
# B = softmax(A, axis=0)
#
# Framework axis 0 -> NumPy axis 1
#
# Therefore NumPy:
#
# A [162,52]
# B [162,52]
# ============================================================

softmax_A = np.random.rand(162, 52)

softmax_B = softmax(softmax_A, axis=1)

# Artificial incoming gradient dL/dB
softmax_dB = np.random.rand(162, 52) - 0.5

# Softmax backward:
#
# dA = B * (
#          dB -
#          sum(B * dB, axis=1, keepdims=True)
#      )

softmax_dA = (
    softmax_B
    * (
        softmax_dB
        - np.sum(
            softmax_B * softmax_dB,
            axis=1,
            keepdims=True
        )
    )
)

# ------------------------------------------------------------
# Write data
# ------------------------------------------------------------

softmax_A.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_1_input.bin"
)

softmax_B.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_1_output.bin"
)

softmax_dB.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_1_incoming_gradient.bin"
)

softmax_dA.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_1_expected_gradient.bin"
)

```

### Softmax gradient Test 2

```python
import numpy as np

np.random.seed(55)


def softmax(x, axis):
    x_max = np.max(x, axis=axis, keepdims=True)
    exp_x = np.exp(x - x_max)
    return exp_x / np.sum(exp_x, axis=axis, keepdims=True)
# ============================================================
# Test 2: Softmax propagation + gradient accumulation
#
# Framework:
#
# A [13,17]
# B = softmax(A, axis=1)
#
# C [72,13]
# E = B @ C
#
# D [13,17]
# F = B + D
#
# L = sum(E) + sum(F)
#
#
# NumPy representation:
#
# A [17,13]
# B [17,13]
# C [13,72]
# E [17,72]
# D [17,13]
# F [17,13]
# ============================================================

prop_A = np.random.rand(17, 13)
prop_C = np.random.rand(13, 72)
prop_D = np.random.rand(17, 13)

# ------------------------------------------------------------
# Forward
# ------------------------------------------------------------

# Framework axis=1 -> NumPy axis=0
prop_B = softmax(prop_A, axis=0)

prop_E = np.matmul(prop_B, prop_C)
prop_F = prop_B + prop_D


# ============================================================
# Backward
# ============================================================

# L = sum(E) + sum(F)

prop_dE = np.ones_like(prop_E)
prop_dF = np.ones_like(prop_F)


# ------------------------------------------------------------
# E = B @ C
# ------------------------------------------------------------

# NumPy:
#
# E = B C
#
# dB = dE C^T
# dC = B^T dE

prop_dB_from_E = np.matmul(
    prop_dE,
    prop_C.T
)

prop_dC = np.matmul(
    prop_B.T,
    prop_dE
)


# ------------------------------------------------------------
# F = B + D
# ------------------------------------------------------------

prop_dB_from_F = prop_dF
prop_dD = prop_dF


# ------------------------------------------------------------
# Gradient accumulation at B
# ------------------------------------------------------------

prop_dB = (
    prop_dB_from_E
    + prop_dB_from_F
)


# ------------------------------------------------------------
# B = Softmax(A, axis=0)
#
# Framework axis=1 -> NumPy axis=0
# ------------------------------------------------------------

prop_dA = (
    prop_B
    * (
        prop_dB
        - np.sum(
            prop_B * prop_dB,
            axis=0,
            keepdims=True
        )
    )
)


# ============================================================
# Write data
# ============================================================

prop_A.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_2_A.bin"
)

prop_C.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_2_C.bin"
)

prop_D.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_2_D.bin"
)

prop_B.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_2_B.bin"
)

prop_E.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_2_E.bin"
)

prop_F.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_2_F.bin"
)

prop_dA.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_2_A_grad.bin"
)

prop_dC.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_2_C_grad.bin"
)

prop_dD.astype(np.float64).tofile(
    "../data/Softmax_Autograd_Test_2_D_grad.bin"
)
```

## TEST_5: Relu incombination with dense layer

### NumPy reference generator

```python
import numpy as np
from pathlib import Path


def relu(x):
    return np.maximum(x, 0.0)


def write_bin(output_dir, filename, values):
    path = output_dir / filename
    np.asarray(values, dtype=np.float64).tofile(path)
    return path


# Dense + ReLU validator
# C++ graph shape:
#   input       : tf_create(tf_float64, 43, 512)
#   dense weight: tf_create(tf_float64, 32, 43)
#   dense bias  : tf_create(tf_float64, 32, 1)
#   dense_layer = dense(32)(input)
#   relu_layer  = relu()(dense_layer)
# NumPy keeps samples on axis 0. C-order flattening matches the C++ tensor
# memory order used by the dense-layer tests: feature/unit + sample * rows.
rng = np.random.default_rng(105)
output_dir = Path("../data")

sample_size = 512
input_features = 43
dense_units = 32
learning_rate = 0.01

x_input = rng.normal(loc=0.0, scale=1.0, size=(sample_size, input_features))
target_output = rng.normal(loc=0.0, scale=1.0, size=(sample_size, dense_units))
weights = rng.normal(loc=0.0, scale=0.15, size=(input_features, dense_units))
bias = rng.normal(loc=0.0, scale=0.25, size=(dense_units,))

# Forward pass: dense(32) -> relu
dense_output = x_input @ weights + bias
relu_output = relu(dense_output)

# Squared-error loss, same per-output-unit reduction style as
# DenseLayer_Loss_Integration.
error = relu_output - target_output
loss = np.mean(error * error, axis=0)

# Backward pass through ReLU into dense.
grad_relu_output = (2.0 / sample_size) * error
grad_dense_output = grad_relu_output * (dense_output > 0.0)
grad_weight = x_input.T @ grad_dense_output
grad_bias = np.sum(grad_dense_output, axis=0)
grad_input = grad_dense_output @ weights.T

updated_weight = weights - learning_rate * grad_weight
updated_bias = bias - learning_rate * grad_bias

files = {
    "input": write_bin(output_dir, "ReluDense_Test_5_input_data.bin", x_input),
    "target_output": write_bin(output_dir, "ReluDense_Test_5_target_output_data.bin", target_output),
    "weight": write_bin(output_dir, "ReluDense_Test_5_weight_data.bin", weights),
    "bias": write_bin(output_dir, "ReluDense_Test_5_bias_data.bin", bias),
    "dense_output": write_bin(output_dir, "ReluDense_Test_5_dense_output_data.bin", dense_output),
    "relu_output": write_bin(output_dir, "ReluDense_Test_5_relu_output_data.bin", relu_output),
    "loss": write_bin(output_dir, "ReluDense_Test_5_loss_data.bin", loss),
    "grad_relu_output": write_bin(output_dir, "ReluDense_Test_5_grad_relu_output_data.bin", grad_relu_output),
    "grad_dense_output": write_bin(output_dir, "ReluDense_Test_5_grad_dense_output_data.bin", grad_dense_output),
    "grad_weight": write_bin(output_dir, "ReluDense_Test_5_grad_weight_data.bin", grad_weight),
    "grad_bias": write_bin(output_dir, "ReluDense_Test_5_grad_bias_data.bin", grad_bias),
    "grad_input": write_bin(output_dir, "ReluDense_Test_5_grad_input_data.bin", grad_input),
    "updated_weight": write_bin(output_dir, "ReluDense_Test_5_updated_weight_data.bin", updated_weight),
    "updated_bias": write_bin(output_dir, "ReluDense_Test_5_updated_bias_data.bin", updated_bias),
}

print("Dense + ReLU TEST_5 reference data generated")
print(f"input: {x_input.shape}, weight: {weights.shape}, bias: {bias.shape}")
print(f"dense_output: {dense_output.shape}, relu_output: {relu_output.shape}")
print(f"loss: {loss.shape}, grad_weight: {grad_weight.shape}, grad_bias: {grad_bias.shape}")
for name, path in files.items():
    print(f"{name}: {path}")
```

## Test 2, DenseLayer_Loss_Integration_parameter_update_multiple_epochs

```python
from dataclasses import dataclass
from typing import Dict
import numpy as np


@dataclass
class DenseTest4SimulateEpoch:
    x: np.ndarray  # (sample_size, in_features)
    w: np.ndarray  # (in_features, dense_units)
    b: np.ndarray  # (dense_units,)
    y: np.ndarray  # (sample_size, dense_units)


def load_dense_test_4(
    sample_size: int,
    no_of_input_features: int,
    no_of_dense_units: int,
    seed: int = 77,
) -> DenseTest4SimulateEpoch:
    rng = np.random.default_rng(seed)
    x = rng.random((sample_size, no_of_input_features))
    w = rng.random((no_of_input_features, no_of_dense_units))
    b = rng.random((no_of_dense_units,))
    y = rng.random((sample_size, no_of_dense_units))
    return DenseTest4SimulateEpoch(x=x, w=w, b=b, y=y)


def simulate_training(
    data: DenseTest4SimulateEpoch,
    epochs: int,
    batch_size: int,
    learning_rate: float,
) -> Dict[str, np.ndarray]:
    x, w, b, y = data.x, data.w, data.b, data.y

    if epochs <= 0:
        raise ValueError("epochs must be > 0")
    if batch_size <= 0 or batch_size > x.shape[0]:
        raise ValueError(f"batch_size must be in [1, {x.shape[0]}]")
    if x.shape[0] != y.shape[0]:
        raise ValueError("x and y must have same sample size")

    sample_size = x.shape[0]
    num_batches = sample_size // batch_size  # drop remainder
    if num_batches == 0:
        raise ValueError("batch_size is too large; no full batch available")

    all_batch_index = []
    all_input_batch = []
    all_output_batch = []
    all_loss_batch = []
    all_grad_loss = []
    all_grad_weight = []
    all_grad_bias = []
    all_updated_weight = []
    all_updated_bias = []
    all_target_output = []

    curr_w = w.copy()
    curr_b = b.copy()

    # One batch per epoch-step, rolling over full batches in order.
    for ep in range(epochs):
        epoch_loss = 0.0
        for batch_idx in range(num_batches):
            start = batch_idx * batch_size
            end = start + batch_size

            x_batch = x[start:end, :]  # (batch_size, in_features)
            y_batch = y[start:end, :]  # (batch_size, dense_units)

            output = x_batch @ curr_w + curr_b
            diff = output - y_batch
            loss = np.mean(diff * diff, axis=0)
            epoch_loss += loss

            grad_loss = (2.0 / batch_size) * diff
            grad_weight = x_batch.T @ grad_loss
            grad_bias = np.sum(grad_loss, axis=0)

            updated_w = curr_w - learning_rate * grad_weight
            updated_b = curr_b - learning_rate * grad_bias
            print(loss.shape)

            all_batch_index.append(batch_idx)
            all_input_batch.append(x_batch.copy())
            all_output_batch.append(output.copy())
            all_loss_batch.append(loss.copy())
            all_grad_loss.append(grad_loss.copy())
            all_grad_weight.append(grad_weight.copy())
            all_grad_bias.append(grad_bias.copy())
            all_updated_weight.append(updated_w.copy())
            all_updated_bias.append(updated_b.copy())
            all_target_output.append(y_batch.copy())

            # step-wise SGD update continuity
            curr_w = updated_w
            curr_b = updated_b

        epoch_loss /= num_batches

    return {
        "batch_index": np.array(all_batch_index, dtype=np.int64),
        "input": np.stack(all_input_batch, axis=0),
        "output": np.stack(all_output_batch, axis=0),
        "target_output": np.stack(all_target_output, axis=0),
        "loss": np.stack(all_loss_batch, axis=0),
        "grad_loss": np.stack(all_grad_loss, axis=0),
        "grad_weight": np.stack(all_grad_weight, axis=0),
        "grad_bias": np.stack(all_grad_bias, axis=0),
        "updated_weight": np.stack(all_updated_weight, axis=0),
        "updated_bias": np.stack(all_updated_bias, axis=0),
    }

```

## Test 3, double dense layer convergance test

```python
np.random.seed(42)

# -----------------------
# Config
# -----------------------
N = 512
D_in, D_hidden, D_out = 27, 31, 7
batch_size = 32
epochs = 100
lr = 0.01
noise_level = 0.01

# -----------------------
# Data (2-layer ground truth)
# -----------------------
X = np.random.randn(N, D_in)

W1_true = np.random.randn(D_in, D_hidden)
b1_true = np.random.randn(D_hidden)

W2_true = np.random.randn(D_hidden, D_out)
b2_true = np.random.randn(D_out)

Z_true = X @ W1_true + b1_true
Y = Z_true @ W2_true + b2_true

# Add noise
Y += noise_level * np.random.randn(*Y.shape)

# -----------------------
# Model init
# -----------------------
W1 = 0.01 * np.random.randn(D_in, D_hidden)
b1 = np.zeros(D_hidden)


W2 = 0.01 * np.random.randn(D_hidden, D_out)
b2 = np.zeros(D_out)


W1_pretrainning = W1.copy()
b1_pretraining = b1.copy()
W2_pretrainning = W2.copy()
b2_pretraining = b2.copy()

all_batch_index = []
all_input = []
all_output = []
all_loss = []
all_grad_loss = []
all_grad_weight_1 = []
all_grad_bias_1 = []
all_updated_weights_1 = []
training_weights_1_batch = []
training_weights_2_batch = []
all_updated_bias_1 = []
all_grad_weights_2 = []
all_grad_bias_2 = []
all_updated_weights_2 = []
all_updated_bias_2 = []
all_target_output = []


# -----------------------
# Training
# -----------------------
num_batches = N // batch_size

for epoch in range(epochs):
    epoch_loss = 0.0

    for i in range(num_batches):
        training_weights_1_batch.append(W1.copy())
        training_weights_2_batch.append(W2.copy())
        
        s = i * batch_size
        e = s + batch_size

        Xb = X[s:e]
        Yb = Y[s:e]

        # Forward
        Z1 = Xb @ W1 + b1
        Yp = Z1 @ W2 + b2

        # Loss
        sample_loss = np.mean((Yp - Yb) ** 2, axis=1)
        loss = np.mean(sample_loss)
        epoch_loss += loss

        # Backward
        dY =  (2.0 / batch_size) * (Yp - Yb)

        dW2 = Z1.T @ dY
        db2 = np.sum(dY, axis=0)

        dZ1 = dY @ W2.T

        dW1 = Xb.T @ dZ1
        db1 = np.sum(dZ1, axis=0)

        # Update
        W1 -= lr * dW1
        b1 -= lr * db1
        W2 -= lr * dW2
        b2 -= lr * db2

        all_input.append(Xb.copy())
        all_updated_bias_1.append(b1.copy())
        all_updated_bias_2.append(b2.copy())
        all_grad_weights_2.append(dW2.copy())
        all_updated_weights_1.append(W1.copy())
        all_updated_weights_2.append(W2.copy())
        all_grad_loss.append(dY.copy())


    epoch_loss /= num_batches
    all_loss.append(epoch_loss.copy())

    if epoch % 5 == 0:
        print(f"Epoch {epoch:3d} | Loss: {epoch_loss:.6f}")

# -----------------------
# Verification
# -----------------------
Y_pred = (X @ W1 + b1) @ W2 + b2
print("\nFinal MSE:", np.mean((Y_pred - Y) ** 2))

# Effective weights check
W_eff = W1 @ W2
b_eff = b1 @ W2 + b2

W_true_eff = W1_true @ W2_true
b_true_eff = b1_true @ W2_true + b2_true

print("W error:", np.linalg.norm(W_eff - W_true_eff))
print("b error:", np.linalg.norm(b_eff - b_true_eff))
```

```python
# Quadratic function y = x^2
import numpy as np
from pathlib import Path


def relu(x):
    return np.maximum(x, 0.0)


def relu_grad(x):
    return (x > 0.0).astype(np.float64)


def write_bin(output_dir, filename, values):
    path = output_dir / filename
    np.asarray(values, dtype=np.float64).tofile(path)
    return path


rng = np.random.default_rng(2048)
output_dir = Path("../data")

sample_size = 1024
input_features = 1
hidden_units = 8
output_features = 1
batch_size = 32
epochs = 1000
learning_rate = 0.01

# Data set: 1024 scalar inputs and their quadratic targets.
x_input = np.linspace(-1.0, 1.0, sample_size, dtype=np.float64).reshape(sample_size, input_features)
target_output = x_input * x_input

# Model architecture: dense(8) -> relu() -> dense(1).
weights_1 = rng.normal(loc=0.0, scale=0.25, size=(input_features, hidden_units))
bias_1 = np.zeros(hidden_units, dtype=np.float64)
weights_2 = rng.normal(loc=0.0, scale=0.25, size=(hidden_units, output_features))
bias_2 = np.zeros(output_features, dtype=np.float64)

initial_weights_1 = weights_1.copy()
initial_bias_1 = bias_1.copy()
initial_weights_2 = weights_2.copy()
initial_bias_2 = bias_2.copy()

num_batches = sample_size // batch_size
input_batch_history = []
target_output_batch_history = []
predicted_output_batch_history = []
loss_batch_history = []

for epoch in range(epochs):
    for batch_index in range(num_batches):
        start = batch_index * batch_size
        end = start + batch_size

        x_batch = x_input[start:end]
        target_batch = target_output[start:end]

        dense_1_output = x_batch @ weights_1 + bias_1
        relu_output = relu(dense_1_output)
        predicted_output = relu_output @ weights_2 + bias_2

        error = predicted_output - target_batch
        loss = np.mean(error * error)

        grad_predicted_output = (2.0 / batch_size) * error
        grad_weights_2 = relu_output.T @ grad_predicted_output
        grad_bias_2 = np.sum(grad_predicted_output, axis=0)
        grad_relu_output = grad_predicted_output @ weights_2.T
        grad_dense_1_output = grad_relu_output * relu_grad(dense_1_output)
        grad_weights_1 = x_batch.T @ grad_dense_1_output
        grad_bias_1 = np.sum(grad_dense_1_output, axis=0)

        weights_1 -= learning_rate * grad_weights_1
        bias_1 -= learning_rate * grad_bias_1
        weights_2 -= learning_rate * grad_weights_2
        bias_2 -= learning_rate * grad_bias_2

        input_batch_history.append(x_batch.copy())
        target_output_batch_history.append(target_batch.copy())
        predicted_output_batch_history.append(predicted_output.copy())
        loss_batch_history.append(loss)

final_predicted_output = relu(x_input @ weights_1 + bias_1) @ weights_2 + bias_2
final_loss = np.mean((final_predicted_output - target_output) ** 2)

input_batch_history = np.array(input_batch_history, dtype=np.float64)
target_output_batch_history = np.array(target_output_batch_history, dtype=np.float64)
predicted_output_batch_history = np.array(predicted_output_batch_history, dtype=np.float64)
loss_batch_history = np.array(loss_batch_history, dtype=np.float64)

files = {
    "input": write_bin(output_dir, "Quadratic_ReluDense_Test_1_input_data.bin", x_input),
    "target_output": write_bin(output_dir, "Quadratic_ReluDense_Test_1_target_output_data.bin", target_output),
    "initial_weights_1": write_bin(output_dir, "Quadratic_ReluDense_Test_1_initial_weights_1.bin", initial_weights_1),
    "initial_bias_1": write_bin(output_dir, "Quadratic_ReluDense_Test_1_initial_bias_1.bin", initial_bias_1),
    "initial_weights_2": write_bin(output_dir, "Quadratic_ReluDense_Test_1_initial_weights_2.bin", initial_weights_2),
    "initial_bias_2": write_bin(output_dir, "Quadratic_ReluDense_Test_1_initial_bias_2.bin", initial_bias_2),
    "final_weights_1": write_bin(output_dir, "Quadratic_ReluDense_Test_1_final_weights_1.bin", weights_1),
    "final_bias_1": write_bin(output_dir, "Quadratic_ReluDense_Test_1_final_bias_1.bin", bias_1),
    "final_weights_2": write_bin(output_dir, "Quadratic_ReluDense_Test_1_final_weights_2.bin", weights_2),
    "final_bias_2": write_bin(output_dir, "Quadratic_ReluDense_Test_1_final_bias_2.bin", bias_2),
    "input_batch_history": write_bin(output_dir, "Quadratic_ReluDense_Test_1_input_batch_history.bin", input_batch_history),
    "target_output_batch_history": write_bin(output_dir, "Quadratic_ReluDense_Test_1_target_output_batch_history.bin", target_output_batch_history),
    "predicted_output_batch_history": write_bin(output_dir, "Quadratic_ReluDense_Test_1_predicted_output_batch_history.bin", predicted_output_batch_history),
    "loss_batch_history": write_bin(output_dir, "Quadratic_ReluDense_Test_1_loss_batch_history.bin", loss_batch_history),
    "final_predicted_output": write_bin(output_dir, "Quadratic_ReluDense_Test_1_final_predicted_output.bin", final_predicted_output),
}

print("Quadratic dense(8) -> relu -> dense training reference data generated")
print(f"input: {x_input.shape}, target_output: {target_output.shape}")
print(f"batch histories: input {input_batch_history.shape}, predicted_output {predicted_output_batch_history.shape}")
print(f"loss history: {loss_batch_history.shape}, final_loss: {final_loss:.10f}")
print(f"mean squared error: {final_loss:.10f}")
for name, path in files.items():
    print(f"{name}: {path}")
```

```python
# Cubic function y = x^3
import numpy as np
from pathlib import Path


def relu(x):
    return np.maximum(x, 0.0)


def relu_grad(x):
    return (x > 0.0).astype(np.float64)


def write_bin(output_dir, filename, values):
    path = output_dir / filename
    np.asarray(values, dtype=np.float64).tofile(path)
    return path


rng = np.random.default_rng(4096)
output_dir = Path("../data")

sample_size = 1024
input_features = 1
hidden_units = 16
output_features = 1
batch_size = 256
epochs = 1000
learning_rate = 0.01

# Data set: 1024 scalar inputs and their cubic targets.
x_input = np.linspace(-1.0, 1.0, sample_size, dtype=np.float64).reshape(sample_size, input_features)
target_output = x_input ** 3

# Model architecture: dense(16) -> relu() -> dense(1).
weights_1 = rng.normal(loc=0.0, scale=0.25, size=(input_features, hidden_units))
bias_1 = np.zeros(hidden_units, dtype=np.float64)
weights_2 = rng.normal(loc=0.0, scale=0.25, size=(hidden_units, output_features))
bias_2 = np.zeros(output_features, dtype=np.float64)

initial_weights_1 = weights_1.copy()
initial_bias_1 = bias_1.copy()
initial_weights_2 = weights_2.copy()
initial_bias_2 = bias_2.copy()

num_batches = sample_size // batch_size
input_batch_history = []
target_output_batch_history = []
predicted_output_batch_history = []
loss_batch_history = []

for epoch in range(epochs):
    for batch_index in range(num_batches):
        start = batch_index * batch_size
        end = start + batch_size

        x_batch = x_input[start:end]
        target_batch = target_output[start:end]

        dense_1_output = x_batch @ weights_1 + bias_1
        relu_output = relu(dense_1_output)
        predicted_output = relu_output @ weights_2 + bias_2

        error = predicted_output - target_batch
        loss = np.mean(error * error)

        grad_predicted_output = (2.0 / batch_size) * error
        grad_weights_2 = relu_output.T @ grad_predicted_output
        grad_bias_2 = np.sum(grad_predicted_output, axis=0)
        grad_relu_output = grad_predicted_output @ weights_2.T
        grad_dense_1_output = grad_relu_output * relu_grad(dense_1_output)
        grad_weights_1 = x_batch.T @ grad_dense_1_output
        grad_bias_1 = np.sum(grad_dense_1_output, axis=0)

        weights_1 -= learning_rate * grad_weights_1
        bias_1 -= learning_rate * grad_bias_1
        weights_2 -= learning_rate * grad_weights_2
        bias_2 -= learning_rate * grad_bias_2

        input_batch_history.append(x_batch.copy())
        target_output_batch_history.append(target_batch.copy())
        predicted_output_batch_history.append(predicted_output.copy())
        loss_batch_history.append(loss)

final_predicted_output = relu(x_input @ weights_1 + bias_1) @ weights_2 + bias_2
total_mse = np.mean((final_predicted_output - target_output) ** 2)

input_batch_history = np.array(input_batch_history, dtype=np.float64)
target_output_batch_history = np.array(target_output_batch_history, dtype=np.float64)
predicted_output_batch_history = np.array(predicted_output_batch_history, dtype=np.float64)
loss_batch_history = np.array(loss_batch_history, dtype=np.float64)

files = {
    "input": write_bin(output_dir, "Cubic_ReluDense_Test_1_input_data.bin", x_input),
    "output": write_bin(output_dir, "Cubic_ReluDense_Test_1_output_data.bin", target_output),
    "initial_weights_1": write_bin(output_dir, "Cubic_ReluDense_Test_1_initial_weights_1.bin", initial_weights_1),
    "initial_bias_1": write_bin(output_dir, "Cubic_ReluDense_Test_1_initial_bias_1.bin", initial_bias_1),
    "initial_weights_2": write_bin(output_dir, "Cubic_ReluDense_Test_1_initial_weights_2.bin", initial_weights_2),
    "initial_bias_2": write_bin(output_dir, "Cubic_ReluDense_Test_1_initial_bias_2.bin", initial_bias_2),
    "predicted_batch_output": write_bin(output_dir, "Cubic_ReluDense_Test_1_predicted_batch_output.bin", predicted_output_batch_history),
    "batch_loss_history": write_bin(output_dir, "Cubic_ReluDense_Test_1_batch_loss_history.bin", loss_batch_history),
}

print("Cubic dense(16) -> relu -> dense training reference data generated")
print(f"input: {x_input.shape}, target_output: {target_output.shape}")
print(f"batch histories: input {input_batch_history.shape}, predicted_output {predicted_output_batch_history.shape}")
print(f"loss history: {loss_batch_history.shape}")
print(f"total mse: {total_mse:.10f}")
for name, path in files.items():
    print(f"{name}: {path}")
```

## Sin wave fitting y = sin(x), i.e -pi <= x <= pi

```python
# Sin wave fitting y = sin(x), -pi <= x <= pi
import numpy as np
from pathlib import Path


def relu(x):
    return np.maximum(x, 0.0)


def relu_grad(x):
    return (x > 0.0).astype(np.float64)


def write_bin(output_dir, filename, values):
    path = output_dir / filename
    np.asarray(values, dtype=np.float64).tofile(path)
    return path


rng = np.random.default_rng(8192)
output_dir = Path("../data")

sample_size = 1025
input_features = 1
hidden_units_1 = 16
hidden_units_2 = 16
output_features = 1
batch_size = 512
epochs = 1000
learning_rate = 0.01

# Data set: 1025 scalar inputs and their sine targets.
x_input = np.linspace(-np.pi, np.pi, sample_size, dtype=np.float64).reshape(sample_size, input_features)
target_output = np.sin(x_input)

# Model architecture: dense(16) -> relu() -> dense(16) -> relu() -> dense(1).
weights_1 = rng.normal(loc=0.0, scale=0.25, size=(input_features, hidden_units_1))
bias_1 = np.zeros(hidden_units_1, dtype=np.float64)
weights_2 = rng.normal(loc=0.0, scale=0.25, size=(hidden_units_1, hidden_units_2))
bias_2 = np.zeros(hidden_units_2, dtype=np.float64)
weights_3 = rng.normal(loc=0.0, scale=0.25, size=(hidden_units_2, output_features))
bias_3 = np.zeros(output_features, dtype=np.float64)

initial_weights_1 = weights_1.copy()
initial_bias_1 = bias_1.copy()
initial_weights_2 = weights_2.copy()
initial_bias_2 = bias_2.copy()
initial_weights_3 = weights_3.copy()
initial_bias_3 = bias_3.copy()

num_batches = sample_size // batch_size
train_sample_size = num_batches * batch_size
epoch_loss_history = []

for epoch in range(epochs):
    batch_losses = []
    for batch_index in range(num_batches):
        start = batch_index * batch_size
        end = start + batch_size

        x_batch = x_input[start:end]
        target_batch = target_output[start:end]

        dense_1_output = x_batch @ weights_1 + bias_1
        relu_1_output = relu(dense_1_output)
        dense_2_output = relu_1_output @ weights_2 + bias_2
        relu_2_output = relu(dense_2_output)
        predicted_output = relu_2_output @ weights_3 + bias_3

        error = predicted_output - target_batch
        batch_losses.append(np.mean(error ** 2))
        grad_predicted_output = (2.0 / batch_size) * error

        grad_weights_3 = relu_2_output.T @ grad_predicted_output
        grad_bias_3 = np.sum(grad_predicted_output, axis=0)
        grad_relu_2_output = grad_predicted_output @ weights_3.T
        grad_dense_2_output = grad_relu_2_output * relu_grad(dense_2_output)

        grad_weights_2 = relu_1_output.T @ grad_dense_2_output
        grad_bias_2 = np.sum(grad_dense_2_output, axis=0)
        grad_relu_1_output = grad_dense_2_output @ weights_2.T
        grad_dense_1_output = grad_relu_1_output * relu_grad(dense_1_output)

        grad_weights_1 = x_batch.T @ grad_dense_1_output
        grad_bias_1 = np.sum(grad_dense_1_output, axis=0)

        weights_1 -= learning_rate * grad_weights_1
        bias_1 -= learning_rate * grad_bias_1
        weights_2 -= learning_rate * grad_weights_2
        bias_2 -= learning_rate * grad_bias_2
        weights_3 -= learning_rate * grad_weights_3
        bias_3 -= learning_rate * grad_bias_3

    epoch_loss_history.append(np.mean(batch_losses))

last_epoch_mse = epoch_loss_history[-1]

files = {
    "input": write_bin(output_dir, "Sin_ReluDense_Test_1_input_data.bin", x_input),
    "target_output": write_bin(output_dir, "Sin_ReluDense_Test_1_target_output_data.bin", target_output),
    "initial_weights_1": write_bin(output_dir, "Sin_ReluDense_Test_1_initial_weights_1.bin", initial_weights_1),
    "initial_bias_1": write_bin(output_dir, "Sin_ReluDense_Test_1_initial_bias_1.bin", initial_bias_1),
    "initial_weights_2": write_bin(output_dir, "Sin_ReluDense_Test_1_initial_weights_2.bin", initial_weights_2),
    "initial_bias_2": write_bin(output_dir, "Sin_ReluDense_Test_1_initial_bias_2.bin", initial_bias_2),
    "initial_weights_3": write_bin(output_dir, "Sin_ReluDense_Test_1_initial_weights_3.bin", initial_weights_3),
    "initial_bias_3": write_bin(output_dir, "Sin_ReluDense_Test_1_initial_bias_3.bin", initial_bias_3),
    "last_epoch_mse": write_bin(output_dir, "Sin_ReluDense_Test_1_last_epoch_mse.bin", np.array([last_epoch_mse])),
}

print("Sin dense(16) -> relu -> dense(16) -> relu -> dense training reference data generated")
print(f"input: {x_input.shape}, target_output: {target_output.shape}")
print(f"dataset_size: {sample_size}, batch_size: {batch_size}, epochs: {epochs}, trained_samples_per_epoch: {train_sample_size}")
print(f"last epoch mse: {last_epoch_mse:.10f}")
for name, path in files.items():
    print(f"{name}: {path}")
```
